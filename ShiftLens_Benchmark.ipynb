## Validate the published benchmark

Each row represents one detector and object category. The full snapshot covers 2,162 held-out images with three independent scores per image.

```python

import json
import math
import re

dbutils.widgets.text("catalog", "", "Catalog (blank uses current)")
dbutils.widgets.text("schema", "shiftlens_portfolio", "Schema")
catalog = dbutils.widgets.get("catalog").strip() or spark.sql("SELECT current_catalog()").first()[0]
schema = dbutils.widgets.get("schema").strip()
for value in (catalog, schema):
    if not re.fullmatch(r"[A-Za-z_][A-Za-z0-9_]*", value):
        raise ValueError("Choose a catalog and schema with simple SQL identifiers.")
namespace = f"`{catalog}`.`{schema}`"
parsed = json.loads(PAYLOAD)
identity = {(row["category"], row["detector"]) for row in parsed}
if len(parsed) != 36 or len(identity) != 36:
    raise ValueError("Expected one row per detector and 12 categories.")
for row in parsed:
    if not all(math.isfinite(row[x]) for x in ("auroc", "average_precision", "threshold")):
        raise ValueError("Non-finite metrics")
    if not 0 <= row["true_rejects"] <= row["defective_test_parts"]:
        raise ValueError("Invalid positive test counts")
    if not 0 <= row["false_rejects"] <= row["normal_test_parts"]:
        raise ValueError("Invalid negative test counts")
df = spark.createDataFrame([dict(row, run_id=RUN_ID, split_sha256=SPLIT_SHA256) for row in parsed])
df.createOrReplaceTempView("shiftlens_release_incoming")
```

## Persist the snapshot

The merge key is run, category, and detector. Rerunning the same snapshot keeps 36 rows.

```python

spark.sql(f"CREATE SCHEMA IF NOT EXISTS {namespace}")
spark.sql(f"""
CREATE TABLE IF NOT EXISTS {namespace}.benchmark_results (
  run_id STRING, split_sha256 STRING, category STRING, detector STRING, true_rejects BIGINT,
  false_rejects BIGINT, defective_test_parts BIGINT, normal_test_parts BIGINT,
  auroc DOUBLE, average_precision DOUBLE, threshold DOUBLE
) USING DELTA
""")
spark.sql(f"""
MERGE INTO {namespace}.benchmark_results AS target
USING shiftlens_release_incoming AS source
ON target.run_id = source.run_id AND target.category = source.category AND target.detector = source.detector
WHEN NOT MATCHED THEN INSERT *
""")
stored = spark.sql(f"SELECT * FROM {namespace}.benchmark_results").where(f"run_id = '{RUN_ID}'")
assert stored.count() == 36, "Incomplete or duplicate benchmark snapshot"
```

## Read the operating trade-offs

Recall is pooled across held-out images. AUROC is a simple average across categories.

```python

spark.sql(f"""
CREATE OR REPLACE VIEW {namespace}.benchmark_summary AS
SELECT run_id, detector, COUNT(*) AS categories,
       SUM(true_rejects) AS defects_caught, SUM(defective_test_parts) AS defective_images,
       SUM(false_rejects) AS good_images_flagged, SUM(normal_test_parts) AS good_images,
       ROUND(SUM(true_rejects) / SUM(defective_test_parts), 4) AS recall,
       ROUND(1000.0 * SUM(false_rejects) / SUM(normal_test_parts), 1) AS false_rejects_per_1000,
       ROUND(AVG(auroc), 3) AS macro_auroc
FROM {namespace}.benchmark_results
GROUP BY run_id, detector
""")
display(spark.sql(f"SELECT * FROM {namespace}.benchmark_summary WHERE run_id = '{RUN_ID}' ORDER BY macro_auroc DESC"))
display(spark.sql(f"""
  SELECT category, detector, true_rejects AS defects_caught,
         false_rejects AS good_images_flagged, ROUND(auroc, 3) AS auroc
  FROM {namespace}.benchmark_results WHERE run_id = '{RUN_ID}'
  ORDER BY category, auroc DESC
"""))
```
